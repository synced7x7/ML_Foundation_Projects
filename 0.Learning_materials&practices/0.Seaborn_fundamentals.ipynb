#Setup

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

#Import Data

```python
crash_df = sns.load_dataset('car_crashes')
crash_df
```

#Distribution Plots

##Distribution Plot

```python
sns.distplot(crash_df['not_distracted'], kde=False, bins = 25)
```

##Joint Plot

```python
sns.jointplot(x= 'speeding', y='alcohol', data = crash_df, kind='hex')
```

##KDE Plot

```python
sns.kdeplot(crash_df['alcohol'])
```

##Pair Plots

```python
tips_df = sns.load_dataset('tips')

sns.pairplot(tips_df, hue = 'sex', palette = 'Blues')
```

##Rug Plot

```python
sns.rugplot(tips_df['tip'])
```

#Categorical Plots

##Bar Plots

```python
sns.barplot(x='sex', y='total_bill', data = tips_df, estimator=np.median)
```

##Count Plot

```python
sns.countplot(x='sex', data=tips_df)
```

##Box Plot

```python
sns.boxplot(x='day', y='total_bill' , data=tips_df, hue='sex')

plt.legend(loc = 0)
```

##Violin Plot

```python
sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex', split=True)
```

##Strip Plot

```python
plt.figure(figsize=(8,5))

sns.stripplot(x='day', y='total_bill', data=tips_df, jitter=True, hue='sex', dodge=True)
plt.legend(loc = 0)
```

##Swarm Plot

```python
sns.swarmplot(x='day', y='total_bill', data=tips_df)
```

#Matrix Plots

##Heatmaps

```python
plt.figure(figsize=(8,6))
sns.set_context('paper', font_scale=1.4)

crash_mx = crash_df.corr(numeric_only=True)

sns.heatmap(crash_mx, annot=True, cmap='Blues')
```

```python
flights = sns.load_dataset('flights')
print(flights)
flights = flights.pivot_table(index = 'month', columns='year', values='passengers')
print(flights)
sns.heatmap(flights, cmap='Blues', linecolor='white', linewidths=1)
```

##Cluster Map

```python
iris = sns.load_dataset('iris')
species = iris.pop('species')
sns.clustermap(iris)

sns.clustermap(flights, cmap='Blues', standard_scale=True)
```

##Facet Grid

```python
tips_fg = sns.FacetGrid(tips_df, col='time', row='smoker', height=4,
                        aspect=1.3, col_order=['Dinner', 'Lunch'], palette='Set1')
tips_fg.map(plt.scatter, 'total_bill', 'tip')
# tips_fg.map(plt.scatter, 'total_bill', 'tip')
```

##Regression Plot

```python
plt.figure(figsize=(8,6))
sns.set_context('paper', font_scale=1.4)
sns.lmplot(x='total_bill', y='tip', hue='sex', data=tips_df,
           markers=['o', '^'],
           scatter_kws={'s':100, 'linewidths': 0.5, 'edgecolor': 'w'})
```
